### Tasks

**Core concepts covered:** array creation, shape/dimensions, indexing and slicing, vectorized operations, broadcasting, aggregation, Boolean filtering, and reshaping.

```python
marks = np.array([45, 67, 72, 55, 81, 63, 49, 90, 76, 58])

print("Shape:", marks.shape)
print("Average:", marks.mean())
print("Highest:", marks.max())
print("Lowest:", marks.min())
print("Above 60:", marks[marks > 60])
```

```python
subject_marks = np.array([
    [45, 60, 55],
    [67, 72, 70],
    [72, 68, 75],
    [55, 58, 61],
    [81, 85, 79]
])

print("Shape:", subject_marks.shape)
print("Subject-wise average:", subject_marks.mean(axis=0))
```

```python

plt.bar(range(1, 6), subject_marks.mean(axis=1))
plt.xlabel("Student")
plt.ylabel("Average Marks")
plt.title("5 Students Average Marks")
plt.show()
```

**Outcome:** Use NumPy arrays to perform numerical calculations and filtering efficiently.

### Broadcasting

Broadcasting allows NumPy to perform operations on arrays with compatible shapes without manually repeating the smaller array.

**Remember:** compatible dimensions are equal, or one of them is `1`.

```python
marks = np.array([
    [60, 70, 80],
    [50, 65, 75],
    [80, 85, 90]
])

bonus = np.array([5, 2, 3])

updated_marks = marks + bonus
print(updated_marks)
print('marks shape:', marks.shape)
print('bonus shape:', bonus.shape)
```

### Tasks

2. Inspect the dataset.
3. Select required columns.
4. Filter students based on marks.

6. Create a grade column.

```python
df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Simran", "Arjun", "Neha"],
    "Gender": ["M", "F", "M", "F", "M", "F"],
    "Subject": ["AI", "AI", "ML", "ML", "AI", "ML"],
    "Marks": [78, 91, 65, 82, 55, 74]
})

df
```

```python
print(df.head(2))
df.info()
print(df.describe())
```

```python
print(df[["Name", "Marks"]])

print("\nStudents scoring above 70:")
print(df[df["Marks"] > 70])

print("\nSorted by marks:")
print(df.sort_values("Marks", ascending=True))
```

```python
df["Grade"] = np.where(df["Marks"] >= 80, "A",
                  np.where(df["Marks"] >= 60, "B", "C"))

print(df)

print("\nAverage marks by subject:")
print(df.groupby("Subject")["Marks"].mean())
```

> **Important:** The objective is not only to plot the graph. The graph must be interpreted.


```python
# Load Titanic dataset
import seaborn as sns

titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
titanic.isna().sum()
```

```python
# 1. Bar chart – survival by passenger class
survival_class = titanic.groupby("pclass")["survived"].mean()

plt.bar(survival_class.index, survival_class.values)
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.title("Survival Rate by Passenger Class")
plt.show()
```

```python
# 2. Histogram – age distribution
plt.hist(titanic["age"].dropna(), bins=10)
plt.xlabel("Age")
plt.ylabel("Number of Passengers")
plt.title("Age Distribution")
plt.show()
```

```python
sns.kdeplot(titanic["age"].dropna())

plt.xlabel("Age")
plt.ylabel("Density")
plt.title("Age Distribution using KDE")

plt.show()
```

```python
# Histogram
plt.hist(titanic["age"].dropna(), bins=10, density=True)

# KDE curve
sns.kdeplot(titanic["age"].dropna())

plt.xlabel("Age")
plt.ylabel("Density")
plt.title("Age Distribution: Histogram + KDE")

plt.show()
```

```python
# 3. Pie chart – survival distribution
survival_counts = titanic["survived"].value_counts()

plt.pie(survival_counts, labels=["Did not survive", "Survived"], autopct="%1.0f%%")
plt.title("Titanic Survival Distribution")
plt.show()
```

```python
# 4. Scatter plot – age vs fare
plt.scatter(titanic["age"], titanic["fare"], alpha=0.5)
plt.xlabel("Age")
plt.ylabel("Fare")
plt.title("Age vs Fare")
plt.show()
```

```python
# 5. Box plot – fare by passenger class
fare_by_class = [
    titanic.loc[titanic["pclass"] == 1, "fare"].dropna(),
    titanic.loc[titanic["pclass"] == 2, "fare"].dropna(),
    titanic.loc[titanic["pclass"] == 3, "fare"].dropna()
]

plt.boxplot(fare_by_class, labels=["1st", "2nd", "3rd"])
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.title("Fare Distribution by Passenger Class")
plt.show()
```

**Outcome:** Select an appropriate graph, create it using Matplotlib, and explain the information visible in the graph.

## Part B – Preprocessing Titanic

Apply the following:

1. **Missing values**
   - Identify missing values.
   - Fill numerical missing values using an appropriate strategy.
   - Fill categorical missing values using an appropriate strategy.

2. **Encoding**
   - Encode categorical variables such as `sex` and `embarked`.

3. **Feature scaling**
   - Select numerical features.
   - Apply standardization using `StandardScaler`.

4. **Final dataset**
   - Check the shape, data types and missing values after preprocessing.


```python
from sklearn.preprocessing import StandardScaler
titanic_prep = titanic.copy()
# Missing values
titanic_prep["age"] = titanic_prep["age"].fillna(titanic_prep["age"].median())
titanic_prep["embarked"] = titanic_prep["embarked"].fillna(titanic_prep["embarked"].mode()[0])
# Encoding
titanic_prep = pd.get_dummies(
    titanic_prep,
    columns=["sex", "embarked"],
    drop_first=True
)
# Scaling selected numerical features
scaler = StandardScaler()
titanic_prep[["age", "fare"]] = scaler.fit_transform(
    titanic_prep[["age", "fare"]]
)
print(titanic_prep.head())
print("\nRemaining missing values:")
print(titanic_prep.isnull().sum())
```

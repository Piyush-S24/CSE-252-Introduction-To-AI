## 1 &middot; Installing what you need

Python code almost never works alone - it relies on libraries other people
wrote. `pip` is the tool that installs them.

That is the same command Practical 1 tells you to fall back on if your
environment check fails. Once it finishes, come back here and run the next
cell.

````python
import sys, importlib, warnings
warnings.filterwarnings("ignore")

missing = []
for name in ["numpy", "pandas", "matplotlib", "sklearn"]:
    try:
        mod = importlib.import_module(name)
        print("%-12s %s" % (name, getattr(mod, "__version__", "ok")))
    except ImportError:
        missing.append("scikit-learn" if name == "sklearn" else name)
        print("%-12s MISSING" % name)

print("\nPython", sys.version.split()[0])

if missing:
    print("\nInstall these first:  pip install " + " ".join(missing))
else:
    print("\n✓ environment ready - this is exactly what Practical 1's first cell checks")
````

````python
def check(label, condition, hint=""):
    """Instant feedback, so you are never stuck waiting for a demonstrator."""
    mark = "✓" if condition else "✗"
    print(mark, label, ("  - hint: " + hint) if (hint and not condition) else "")
    return bool(condition)

print("✓ checker ready")
````

## 2 &middot; How a notebook remembers things

A notebook remembers **what you ran**, not what you can see on the screen.
Run the next two cells in order, then read what happens.

````python
favourite_number = 7
print("favourite_number set to", favourite_number)
````

````python
print("the kernel still remembers:", favourite_number)
````

The value lives in the notebook's memory (its *kernel*), not in the cell
itself. It stays there even if you delete the cell that created it - which is
how a notebook ends up working for you and failing for everyone else who
opens it fresh.

> **The habit, from day one:** before you show a notebook to anyone - a
> demonstrator, an examiner, a teammate - use **Kernel &rarr; Restart & Run
> All**. That proves it actually works from a clean start. You will see this
> exact advice again in Practical 1.

## 4 &middot; A small preview of what Practical 1 looks like

Practical 1 does exactly this pattern, just with NumPy arrays instead of a
plain list, and with random sensor noise added in. Nothing new to learn here
- just recognise the shape of it.

````python
# a tiny "sensor log" as a plain list, no library needed
readings = [2.1, 2.3, 1.9, 2.0, 2.2, 2.4, 2.1]

total = 0
for r in readings:
    total = total + r
mean_reading = total / len(readings)

print("readings:", readings)
print("mean reading: %.2f" % mean_reading)
````
